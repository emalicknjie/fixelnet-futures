# Fixelnet Runner
Thin notebook that calls the `fixelnet` package. All logic lives in `fixelnet/`.

```python
import warnings
warnings.simplefilter('ignore')

import seaborn as sns
sns.set_style('darkgrid')
sns.set_palette('muted')
sns.set_context('notebook', font_scale=1.5, rc={'lines.linewidth': 2.5})

%matplotlib inline
```

```python
from fixelnet import load_model, run_predictions, compute_signal_stats

MODEL_DIR  = '/home/emalick/Desktop/Trading/nnet_state_saves/test'
OUTPUT_DIR = '/home/emalick/Desktop/Trading/fixelnet_server'
YEAR       = 2025
```

```python
model = load_model(MODEL_DIR)
```

```python
results = run_predictions(
    model=model,
    year=YEAR,
    output_dir=OUTPUT_DIR,
)
```

```python
for asset, df in results.items():
    stats = compute_signal_stats(df)
    print(f"{asset}: Long {stats['buy_pct']}%  HeavyShort {stats['sell_pct']}%  Short {stats['hold_pct']}%")
```
